```python
import pygame
import random

pygame.init()


screen_width = 600
screen_height = 600


screen = pygame.display.set_mode((screen_width, screen_height))
pygame.display.set_caption("Игра 'Ловить мячи'")


pygame.display.flip()

WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
RED = (255, 0, 0)

item = pygame.image.load('C:/Imagnes_files/Bear_g.webp')
cart_image = pygame.image.load('C:/Imagnes_files/Telega')




score = 0


font = pygame.font.SysFont(None, 30)


cart_x = 250
cart_y = 500
cart_speed = 0


def create_item():
    item_x = random.randint(0, screen_width - 40)
    item_y = -50
    item_speed = random.randint(1, 5)
    return {'x': item_x, 'y': item_y, 'speed': item_speed}



items = []


running = True
while running:
    # Обработка событий (нажатия клавиш, закрытие окна и т.д.)
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.KEYDOWN:
            if event.key == pygame.K_LEFT:
                cart_speed = -5
            elif event.key == pygame.K_RIGHT:
                cart_speed = 5
        elif event.type == pygame.KEYUP:
            if event.key == pygame.K_LEFT and cart_speed < 0:
                cart_speed = 0
            elif event.key == pygame.K_RIGHT and cart_speed > 0:
                cart_speed = 0


    screen.fill(BLACK)

    # Добавление новых предметов каждые 30 кадров
    if len(items) < 10 and pygame.time.get_ticks() % 30 == 0:
        new_item = create_item()
        items.append(new_item)

    # Движение тележки
    cart_x += cart_speed
    if cart_x < 0:
        cart_x = 0
    elif cart_x > screen_width - 40:
        cart_x = screen_width - 40


    # Отрисовка предметов (шишек)
    for item in items:
        item['y'] += item['speed']
        if item['y'] > screen_height:
            items.remove(item)
        item_rect = pygame.Rect(item['x'], item['y'], 40, 40)
        screen.fill(RED, item_rect)
        if item_rect.colliderect(pygame.Rect(cart_x, cart_y, 40, 40)):
            items.remove(item)
            score += 10

    score_text = font.render("Score: " + str(score), True, WHITE)
    screen.blit(score_text, (10, 10))


    pygame.display.flip()


pygame.quit()
```
