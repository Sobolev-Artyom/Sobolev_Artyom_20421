```python
class Song:
    def __init__(self, name: str):
        self.name = name
        self.albums = []


class Author:
    def __init__(self, name: str):
        self.name = name
        self.albums = []

    def add(self, album):
        self.albums.append(album)
        album.authors.append(self)


class Album:
    def __init__(self, name: str, date: int):
        self.name = name
        self.date = date
        self.authors = []
        self.songs = []

    def add(self, song):
        self.songs.append(song)
        song.albums.append(self)


class PlayList:
    def __init__(self, name: str):
        self.name = name
        self.songs = []

    def add(self, song):
        self.songs.append(song)

first_author = Author('КиШ')
second_author = Author('Imagne Dragons')

first_album = Album('example', 1985)
second_album = Album('s_example', 2013)

song_1 = Song('Лесник')
song_2 = Song('Beliver')

first_author.add(first_album)
second_author.add(second_album)

first_album.add(song_1)
second_album.add(song_2)

play_list = PlayList('Rock')
play_list.add(song_1)
```
