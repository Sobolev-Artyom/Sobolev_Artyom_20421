```python
class Users:
    def __init__(self):
        self.users =[]

    def add(self, storage):
        for u in storage:
            try: self.check(u)
            except(InvalidUs_NameExeption,InvalidAgeException,LowageException,InvalidEmailDataException) as e:
                print(f'Попытка добавления отклонена,{u}: {e}')
            self.users.append((u[0], u[1]))

    def check(self, user):
        if user[0] in [i[0] for i in self.users]:
            raise InvalidUs_NameExeption(f"Имя пользователя {user[0]} не уникально")
        if type(user[2]) != int or user[2] < 0:
            raise InvalidAgeException(f"Возраст не является положительным целым числом")
        if user[2] < 16:
            raise LowageException(f"Пользователю меньше 16 лет")
        if user[1].count("@") != 1 or user[1].find("@") == 0 and user[1].find("@") == len(user[1]) - 1:
            raise InvalidEmailDataException(f"Адрес электронной почты недействителен ")


class InvalidUs_NameExeption(Exception):
    pass

class InvalidAgeException(Exception):
    pass

class LowageException(Exception):
    pass

class InvalidEmailDataException(Exception):
    pass

data = [("Andrey", "9234743.ry", 18),
        ("Daniil", "pskovskiy@.com", 28),
        ("Teo", "teo@example@.com", 40),
        ("Mark", "mir@example@.com", -10),
        ("Any", "any@@example@.com", 9)]

uc = Users()
uc.add(data)
```
