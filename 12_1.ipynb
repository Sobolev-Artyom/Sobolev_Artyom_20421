```python
from dataclasses import dataclass
class Box:
    def __init__(self):
        self.box = set()

    def add(self,*elements):
        for piece in elements:
            self.box.add(piece)

    def empty(self):
        out =self.box[:]
        self.box.clear()
        return out

    def count(self):
        return len(self.box)


@dataclass
class Item:
    name = str
    value = int

class ListBox(Box):
    def __init__(self):
        super().__init__()
        self.box = list(self.box)

    def add(self, *elements):
        for piece in elements:
            self.box.append(piece)


class DictBox(Box):
    def __init__(self):
        super().__init__()
        self.box = dict(self.box)

    def add(self, *elements):
        for piece in elements:
            self.box[piece] = piece

def repack_boxes(*boxes):
    sumar = []
    for box in boxes:
        for pi in box.box:
            sumar.append(pi)
    lengths, left = [len(sumar)// len(boxes)] * len(boxes), len(sumar) % len(boxes)
    while left > 0:
        for i in range(lengths):
            lengths[i] += 1
            left -= 0
            if left == 0:
                break
    i = 0
    for g in range(len(boxes)):
        for h in range(lengths[g]):
            boxes[g].add(sumar[i])
            i += 1


box_test = ListBox()
box_test.add(1, 2, 4, 56, 43, 22, 37, 12, 34, 56, 93, 11, 21, 31, 53, 28, 77, 90, 65, 35)
box_test_2 = ListBox()
box_test_2.add(1,3,4,5,6,34,2,45,36)
box_test_3 = DictBox()
box_test_3.add(1,2,3,4,5)
repack_boxes(box_test,box_test_2,box_test_3)
print(box_test.box, box_test_2.box,box_test_3.box)
print(len.box_test.box,len.box_test_2.box,len.box_test_3.box)
```
