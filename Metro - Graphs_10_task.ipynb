```python
class Vertex:
    def __init__(self):
        self._links = []

    @property
    def links(self) -> list["Link"]:
        return self._links[:]
    
class Link:
    def __init__(self, v1: Vertex, v2: Vertex, dist: int = 1):
        self._v1 = v1
        self._v2 = v2
        self._dist = dist

    @property
    def v1(self):
        return self._v1

    @property
    def v2(self):
        return self._v2

    @property
    def dist(self):
        return self._dist
    
    @dist.setter
    def dist(self, value):
        self._dist = value

    def __eq__(self, other):
        return set([self.v1, self.v2]) == set([other.v1, other.v2])
    
class LinkedGraph:
    def __init__(self):
        self._vertex = []
        self._links = []

    def add_vertex(self, v: Vertex):
        if not list(filter(lambda x: x == v, self._vertex)):
            self._vertex.append(v)

    def add_link(self, lnk: Link):
        if not list(filter(lambda x: x == lnk, self._links)):
            self._links.append(lnk)
            self.add_vertex(lnk.v1)
            self.add_vertex(lnk.v2)
            lnk.v1._links.append(lnk)
            lnk.v2._links.append(lnk)

        # инициализация списка расстояний для каждой вершины
    def find_path(self, start_v: Vertex, stop_v: Vertex) -> tuple[list[Vertex], list[Link]]:
        distances = {v: float('inf') for v in self._vertex}
        distances[start_v] = 0

        # инициализация списка посещенных вершин
        visited = set()

        #  проведение поиска кратчайшего пути
        while True:
            cur_dist = float('inf')
            cur_v = None

            # нахождение вершины с минимальным значением расстояния из невзятых вершин
            for v in set(self._vertex) - visited:
                if distances[v] < cur_dist:
                    cur_v = v
                    cur_dist = distances[v]

            if cur_v is None:
            # после полной обработки вершин, выходим из цикла
                break

            if cur_v == stop_v:
                # если нашли нужную вершину, возвращаем список вершин и связей на пути
                path_vertices = []
                path_links = []
                while cur_v != start_v:
                    prev_v, link = min(((v, lnk) for lnk in cur_v.links for v in (lnk.v1, lnk.v2) if v != cur_v and distances[v] == distances[cur_v] - lnk.dist), key=lambda x: distances[x[0]])
                    path_vertices.append(cur_v)
                    path_links.append(link)
                    cur_v = prev_v
                path_vertices.append(start_v)
                return path_vertices[::-1], path_links[::-1]

            visited.add(cur_v)

            # обновляем расстояния до соседних вершин
            for link in cur_v.links:
                next_v = link.v1 if link.v2 == cur_v else link.v2
                new_dist = cur_dist + link.dist
                if new_dist < distances[next_v]:
                    distances[next_v] = new_dist

        return [], []


class Station(Vertex):
    def __init__(self, name: str):
        super().__init__()
        self.name = name
        
    def __str__(self) -> str:
        return self.__repr__()
    
    def __repr__(self) -> str:
        return self.name

class LinkMetro(Link):
    def __init__(self, v1: Vertex, v2: Vertex, dist: int  = 1):
        super().__init__(v1, v2, dist)   


map2 = LinkedGraph()
v1 = Vertex()
v2 = Vertex()
v3 = Vertex()
v4 = Vertex()
v5 = Vertex()
map2.add_link(Link(v1, v2))
map2.add_link(Link(v2, v3))
map2.add_link(Link(v2, v4))
map2.add_link(Link(v3, v4))
map2.add_link(Link(v4, v5))
assert len(map2._links) == 5, "неверное число связей в списке _links класса LinkedGraph"
assert len(map2._vertex) == 5, "неверное число вершин в списке _vertex класса LinkedGraph"
map2.add_link(Link(v2, v1))
assert len(map2._links) == 5, "метод add_link() добавил связь Link(v2, v1), хотя уже имеется связь Link(v1, v2)"
path = map2.find_path(v1, v5)
s = sum([x.dist for x in path[1]])
assert s == 3, "неверная суммарная длина маршрута, возможно, некорректно работает объект-свойство dist"
assert issubclass(Station, Vertex) and issubclass(LinkMetro, Link), "класс Station должен наследоваться от класса Vertex, а класс LinkMetro от класса Link"
map2 = LinkedGraph()
v1 = Station("1")
v2 = Station("2")
v3 = Station("3")
v4 = Station("4")
v5 = Station("5")
map2.add_link(LinkMetro(v1, v2, 1))
map2.add_link(LinkMetro(v2, v3, 2))
map2.add_link(LinkMetro(v2, v4, 7))
map2.add_link(LinkMetro(v3, v4, 3))
map2.add_link(LinkMetro(v4, v5, 1))
assert len(map2._links) == 5, "неверное число связей в списке _links класса LinkedGraph"
assert len(map2._vertex) == 5, "неверное число вершин в списке _vertex класса LinkedGraph"
path = map2.find_path(v1, v5)
assert str(path[0]) == '[1, 2, 3, 4, 5]', path[0]
s = sum([x.dist for x in path[1]])
assert s == 7, "неверная суммарная длина маршрута для карты метро"        


```
