```python
def main():
    def read_file(filename): # Функция для чтения файла
        with open(filename) as f: 
            text = f.read()
        return [int(i) for i in list(text) if i != ' '] # Перебор всех целых чисел .

    
    def write_file(filename, text_to_write): # Функчия записи в файл
        with open(filename, 'w') as f:
            f.write(text_to_write) 

    data = read_file("file_lab_1.txt") # Чтение исходного файла и присвоение переменной data
    
    
    def recursion(idx, N, S, itog_sum, X, sums_sequence, signs): # Рекурсия для программы
        if idx == N-1:
            if itog_sum == S:
                return True 
            else:
                return False # Проверка итоговой суммы и суммы , которую мы ввели (s)
        if itog_sum + X[idx+1] - sums_sequence[idx+2] <= S: # Если итоговая сумма + первое число - последовательность суммы(второе число) <= (s) 
            signs[idx] = '+'
            r = recursion(idx+1, N, S, itog_sum+X[idx+1], X, sums_sequence, signs) # Добавляем знак + и присваиваем рекурсию до р для проверки
            if r:
                return True 
        if itog_sum - X[idx+1] + sums_sequence[idx+2] >= S: # Если итоговая сумма - первое число + последовательность суммы(второе число) >= (s)
            signs[idx] = '-'
            r = recursion(idx+1, N, S, itog_sum - X[idx+1], X, sums_sequence, signs) # Ставим знак - и аналогично
            if r:
                return True 
        return False
    
   
    N = data[0]
    S = data[-1]
    X = data[1:-1] 

    signs = ['']*N # список со знаками
    itog_sum = X[0]
    sums_sequence = [sum(X[i:]) for i in range(N)] + [0] # Перечисление переменных и их функциональностей.

    r = recursion(0, N, S, itog_sum, X, sums_sequence, signs) 
    if r:
        result = ''.join([f'{X[i]}{signs[i]}' for i in range(N)]) + f'={S}' # Собираем выражение воедино
        write_file('result_lab1.txt', result)
    else:
        write_file('result_lab1.txt', 'No solution') # Если это не удается, записываем, что решений нет.


if __name__=="__main__":
    main()    
```
